# PyMuPDF PDF text extraction

Put your PDF in this project folder, then change `pdf_path` below.

```python
from pathlib import Path
import fitz

pdf_path = Path("sample.pdf")  # TODO: change this to your PDF file name

with fitz.open(pdf_path) as doc:
    print(f"pages: {doc.page_count}")
    for page_number, page in enumerate(doc, start=1):
        text = page.get_text("text").strip()
        print(f"\n--- page {page_number} ---")
        print(text[:2000] if text else "(no text found)")
```

```python
# Save all extracted text to a .txt file
out_path = pdf_path.with_suffix(".txt")

with fitz.open(pdf_path) as doc:
    all_text = []
    for page_number, page in enumerate(doc, start=1):
        all_text.append(f"--- page {page_number} ---")
        all_text.append(page.get_text("text").strip())

out_path.write_text("\n\n".join(all_text), encoding="utf-8")
out_path
```
